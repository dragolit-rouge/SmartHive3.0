```python
%%sql
CREATE TABLE IF NOT EXISTS pasek (
    id INTEGER PRIMARY KEY,
    id_uley INTEGER NOT NULL
    );
```

```python
%%sql
CREATE TABLE IF NOT EXISTS uley (
    id INTEGER NOT NULL,
    id_sensor INTEGER NOT NULL,
    FOREIGN KEY (id) REFERENCES pasek(id_uley)
        ON UPDATE CASCADE
        ON DELETE SET NULL,
    PRIMARY KEY(id, id_sensor)
);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS sensor (
    id INTEGER NOT NULL,
    rtime TEXT NOT NULL DEFAULT('now'),
    date REAL NOT NULL DEFAULT(0),
    units TEXT NOT NULL,
    FOREIGN KEY (id_sensor) REFERENCES uley(id_sensor)
        ON UPDATE CASCADE
        ON DELETE SET NULL
    );
```
