```python
%%sql
CREATE TABLE IF NOT EXISTS pasek (
    id INTEGER PRIMARY KEY,
    name TEXT
    );
```

```python
%%sql
CREATE TABLE IF NOT EXISTS uley (
    id INTEGER UNIQUE NOT NULL,
    id_pasek INTEGER,
--сначала прописываем все параметры - потом прописываем наследование
    FOREIGN KEY (id_pasek) REFERENCES pasek(id),
--какой параметр в этой таблице мы берем из другой
    PRIMARY KEY(id, id_pasek)
--уникальное сочетание
    );
```

```python
%%sql
CREATE TABLE IF NOT EXISTS sensor (
    id INTEGER UNIQUE NOT NULL,
    id_uley INTEGER,
    rtime TEXT NOT NULL DEFAULT('now'),
    date REAL NOT NULL DEFAULT(0),
    units TEXT NOT NULL,
    FOREIGN KEY (id_uley) REFERENCES uley(id)
    );
```
